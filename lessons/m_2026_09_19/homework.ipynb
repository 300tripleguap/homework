# Часть 1. Работа над ошибками

Прошлая домашка сдана, все проверки зелёные. Но в пяти местах решение сработало
случайно или не тем способом. Разбираем - это важнее, чем правильные ответы.

Примеры «как правильно» специально даны на **других данных**, чтобы не подсказать
ответы к контрольной во второй части.

### Ошибка 1. Сравнение чисел через текст

**Было:**

```python
is_equal = value = f"{value:.1f}" == "0.3"
```

**Что не так - две вещи сразу:**

1. Сравниваются не числа, а **строки**: `"0.3" == "0.3"`. Ответ совпал, но задача была про числа.
   На других данных такой приём подведёт: `f"{2.0:.1f}"` даст `"2.0"`, и сравнение с `"2"` вернёт `False`,
   хотя числа равны.
2. `is_equal = value = ...` - это **двойное присваивание**: результат сравнения записался
   и в `is_equal`, и в `value`. Исходное число из `value` пропало. Если ниже по коду оно
   понадобится - получишь `True` вместо числа и странную ошибку.

**Как правильно** (числа сравниваем как числа):

```python
x = 0.7 + 0.1          # 0.7999999999999999
print(x == 0.8)         # False - прямое сравнение float ненадёжно
print(round(x, 2) == 0.8)   # True - округлили оба до нужной точности
```

### Ошибка 2. `and` и `or` без скобок

**Было:**

```python
can_ride = (age >= 12 and height >= 140 or with_adult)
```

**Что не так:** ответ совпал случайно. `and` выполняется **раньше** `or` - как умножение раньше
сложения в математике. Поэтому Python прочитал это так:

```python
(age >= 12 and height >= 140) or with_adult
```

В той задаче это совпало с нужным смыслом. Но стоит поменять правило местами - и логика развалится:

```python
# "пускают участников и тренеров, но только с пропуском"
is_member, is_coach, has_pass = True, False, False

print(is_member or is_coach and has_pass)      # True  - Python понял: is_member or (is_coach and has_pass)
print((is_member or is_coach) and has_pass)    # False - а нужно было так
```

**Правило:** если в условии есть и `and`, и `or` - ставь скобки всегда, даже когда кажется,
что и так понятно. Скобки бесплатны, отладка - нет.

### Ошибка 3. `None` и `0` - это не одно и то же

**Было:**

```python
text_b = "скидки нет" if discount_b == 0 else discount_b
```

**Что не так:** проверялось `== 0`, а задача была про различие «данных нет» и «значение равно нулю».
Такая проверка не заметит `None`, зато сработает на честном нуле.

**Почему это важно** - пример с датчиком температуры:

```python
temp_a = None    # датчик не ответил, температура неизвестна
temp_b = 0       # датчик ответил: ровно ноль градусов

print(temp_a is None)    # True  - данных нет
print(temp_b is None)    # False - данные есть, и это ноль

# А вот так делать нельзя - оба случая сольются в один:
print(bool(temp_a), bool(temp_b))   # False False
```

**Правило:** отсутствие значения проверяется только через `is None` / `is not None`.
Не `== None`, не `if x:`, не `== 0`.

### Ошибка 4. Переменная из чужой ячейки

**Было:**

```python
ts = time.time()
now_text = (dt.strftime("%d.%m.%Y %H:%M"))
```

**Что не так:** переменной `dt` в этой ячейке никто не создавал - она осталась от примера выше.
Код запустился только потому, что блокнот выполнялся сверху вниз. Последствия:

- в `now_text` попало время **из примера**, а не из твоего `ts` - задача решена не та;
- если открыть блокнот заново и запустить одну эту ячейку, будет `NameError`.

**Как правильно:** всё, что нужно ячейке, в ней и создаём.

```python
import time
from datetime import datetime

ts = time.time()
dt = datetime.fromtimestamp(ts)       # свой dt, полученный из своего ts
print(dt.strftime("%d.%m.%Y %H:%M"))
```

### Ошибка 5. Не тот инструмент для списка

**Было:**

```python
del shopping[1]
shopping.insert(1, 'яйца')
shopping.insert(2, 'сыр')
```

**Что не так:** результат верный, но способ неудобный. В задании было «добавь в конец»
и «удали молоко» - для этого есть отдельные методы:

| Что нужно | Правильный метод | Что делал ты |
|---|---|---|
| Добавить в конец | `append("сыр")` | `insert(2, "сыр")` - пришлось считать индекс |
| Удалить по значению | `remove("молоко")` | `del shopping[1]` - пришлось искать индекс глазами |

Разница проявится на длинном списке: после каждой вставки все индексы справа сдвигаются,
и держать их в голове становится невозможно. Подробно разберём в Части 3.

### Проверь сам

Запусти ячейку и убедись, что всё написанное выше - правда.

````python
# 1. Сравнение чисел через текст подводит
x = 2.0
print(f"{x:.1f}" == "2", "<- текстом", "|", round(x, 2) == 2, "<- числами")

# 2. Приоритет and над or
is_member, is_coach, has_pass = True, False, False
print(is_member or is_coach and has_pass, "<- без скобок", "|",
      (is_member or is_coach) and has_pass, "<- со скобками")

# 3. None против нуля
temp_a = None
temp_b = 0
print(temp_a is None, temp_b is None, "<- is None различает", "|",
      bool(temp_a), bool(temp_b), "<- а bool() сливает в одно")
````

# Часть 3. Новая тема: списки - `list`

## Описание

**Список - это несколько значений, которые лежат под одним именем.**
Записывается в квадратных скобках через запятую:

```python
marks = [5, 3, 4, 5, 2]
```

### Зачем он нужен

Представь, что надо хранить оценки за четверть. Без списка пришлось бы завести
`mark1`, `mark2`, `mark3`... и так до конца четверти, а потом руками складывать их все.
Со списком это одно имя и одна команда `sum(marks)`.

Список нужен, когда:

- значений много и они однотипные (оценки, товары, имена);
- заранее неизвестно, сколько их будет (корзина в магазине то пустая, то на 20 позиций);
- важен **порядок** (очередь, шаги рецепта, история сообщений).

### Три свойства списка

| Свойство | Что значит |
|---|---|
| Упорядоченный | У каждого элемента есть индекс: `marks[0]` - первая оценка. Нумерация с нуля |
| Изменяемый | Можно добавлять, удалять и заменять элементы прямо в этом же списке |
| Хранит что угодно | Числа, строки, `None`, даже другие списки - и всё вперемешку |

### Список и строка - близкие родственники

У них одинаково работают `len()`, индексы, срезы и проверка `in`. Разница в одном,
но принципиальном:

```python
word = "кот"
word[0] = "р"       # ОШИБКА: строку менять нельзя

animals = ["кот"]
animals[0] = "пёс"  # работает: список менять можно
```

Поэтому методы строк **возвращают новую строку** (`name = name.upper()`),
а методы списков **меняют сам список** (`animals.append("пёс")`, без присваивания).
Это главный источник путаницы - держи в голове.

## Примеры

### Пример 1. Создаём и читаем

````python
marks = [5, 3, 4, 5, 2]
basket = []                             # пустой список - нормальное начало
mixed = [1, "текст", 3.5, True, None]   # типы можно мешать

print(marks)
print(len(marks))     # 5 - сколько элементов
print(marks[0])       # 5 - первый (нумерация с нуля)
print(marks[-1])      # 2 - последний
print(marks[1:3])     # [3, 4] - срез, правая граница не входит
print(3 in marks)     # True - есть ли такое значение
print(len(basket))    # 0

# Готовые функции для чисел
print(sum(marks))     # 19
print(max(marks))     # 5
print(min(marks))     # 2
print(sum(marks) / len(marks))   # 3.8 - средний балл
````

### Пример 2. Меняем список

Обрати внимание: у этих методов **нет присваивания**. Они меняют список на месте
и возвращают `None` - если написать `marks = marks.append(5)`, список потеряется.

````python
cars = ["bmw", "audi"]

cars.append("volvo") # добавить в конец
print(cars)                 # ['bmw', 'audi', 'volvo']

cars.insert(1, "kia")       # вставить по индексу, остальные сдвигаются вправо
print(cars)                 # ['bmw', 'kia', 'audi', 'volvo']

cars[0] = "lada"            # заменить по индексу
print(cars)                 # ['lada', 'kia', 'audi', 'volvo']

cars.remove("kia")          # удалить по значению (первое совпадение)
del cars[0]                 # удалить по индексу
last = cars.pop()           # удалить последний И вернуть его значение
print(cars, "| достали:", last)

# Сортировка: два варианта, не путать
numbers = [7, 2, 9]
print(sorted(numbers))      # [2, 7, 9] - НОВЫЙ список, исходный цел
print(numbers)              # [7, 2, 9]
numbers.sort()              # сортирует сам список, возвращает None
print(numbers)              # [2, 7, 9]
numbers.reverse()           # перевернуть
print(numbers)              # [9, 7, 2]
````

### Пример 3. Списки и строки, вложенность и коварное копирование

````python
# Строка <-> список
print("хлеб,молоко,сыр".split(","))      # ['хлеб', 'молоко', 'сыр']
print(", ".join(["хлеб", "молоко"]))     # хлеб, молоко

# Список внутри списка: доступ в два шага
table = [["Егор", 15], ["Аня", 14]]
print(table[0])        # ['Егор', 15]
print(table[0][0])     # Егор - сначала строка таблицы, потом ячейка

# ЛОВУШКА: присваивание НЕ создаёт копию, это второе имя для того же списка
a = [1, 2, 3]
b = a
b.append(4)
print(a)       # [1, 2, 3, 4] - изменился и a, хотя трогали b!

# Настоящая копия
a = [1, 2, 3]
b = a.copy()   # или b = a[:]
b.append(4)
print(a, b)    # [1, 2, 3] [1, 2, 3, 4] - теперь списки независимы
````

## Что дальше: перебор

Пока мы обращаемся к элементам поштучно - `marks[0]`, `marks[1]`. Это работает, пока
элементов три. А если их триста?

Для этого есть цикл `for` - «пройди по списку и сделай что-то с каждым элементом».
Разберём его на занятии, а пока просто посмотри, как это выглядит:

```python
for mark in marks:
    print(mark)
```

Читается дословно: «для каждой оценки `mark` из списка `marks` - напечатать её».

````python
marks = [5, 3, 4, 5, 2]

# Поштучно - работает, но так нельзя писать всерьёз
print(marks[0])
print(marks[1])
print(marks[2])

print("---")

# То же самое циклом - и список может быть любой длины
for mark in marks:
    print(mark)
````

### Задача 5. Оценки за четверть

Дан список `marks`. Посчитай:

- `ordered` - оценки по возрастанию, **исходный список менять нельзя**;
- `top3` - три лучшие оценки, по убыванию;
- `average` - средний балл, округлённый до двух знаков;
- `backup` - **независимая копия** `marks`.

Проверка сама добавит в `backup` пятёрку и убедится, что `marks` от этого не изменился.

````python
marks = [4, 5, 3, 5, 2]

# ТВОЙ КОД
ordered = sorted(marks)
top3 = sorted(marks, reverse=True)
top3 = top3[0:3]
average = sum(marks) / len(marks) 
backup = marks.copy()
print(ordered, top3, average)
print(marks, backup)
````

````python
# Проверка
assert marks == [4, 5, 3, 5, 2], (
    f"исходный marks изменился: {marks}. "
    "Проверь, чем ты сортировал и как делал копию"
)
assert ordered == [2, 3, 4, 5, 5], f"ожидалось [2, 3, 4, 5, 5], а вышло {ordered}"
assert top3 == [5, 5, 4], f"ожидалось [5, 5, 4], а вышло {top3}"
assert average == 3.8, f"ожидалось 3.8, а вышло {average}"
assert isinstance(backup, list), "backup должен быть списком"
backup.append(5)
assert marks == [4, 5, 3, 5, 2], (
    f"после добавления в backup исходный marks стал {marks}. "
    "Значит, backup - это не копия, а второе имя того же списка"
)
assert backup == [4, 5, 3, 5, 2, 5], f"backup должен быть копией marks, а вышло {backup}"
print("OK - задача 5 решена")
````

## Памятка по спискам

`lst = [10, 20, 30]`

| Что нужно | Как | Результат | Меняет список? |
|---|---|---|---|
| Длина | `len(lst)` | `3` | нет |
| Первый / последний | `lst[0]` / `lst[-1]` | `10` / `30` | нет |
| Кусок списка | `lst[0:2]` | `[10, 20]` | нет |
| Есть ли значение | `20 in lst` | `True` | нет |
| Сумма / макс / мин | `sum(lst)` / `max(lst)` / `min(lst)` | `60` / `30` / `10` | нет |
| Добавить в конец | `lst.append(40)` | `None` | **да** |
| Вставить по индексу | `lst.insert(0, 5)` | `None` | **да** |
| Заменить по индексу | `lst[0] = 99` | - | **да** |
| Удалить по значению | `lst.remove(20)` | `None` | **да** |
| Удалить по индексу | `del lst[0]` | - | **да** |
| Удалить и забрать | `lst.pop()` | `30` | **да** |
| Отсортировать на месте | `lst.sort()` | `None` | **да** |
| Получить отсортированный | `sorted(lst)` | новый список | нет |
| По убыванию | `sorted(lst, reverse=True)` | новый список | нет |
| Перевернуть | `lst.reverse()` | `None` | **да** |
| Копия | `lst.copy()` | новый список | нет |
| Строка -> список | `"a,b".split(",")` | `['a', 'b']` | - |
| Список -> строка | `", ".join(['a', 'b'])` | `"a, b"` | - |

**Колонка «Результат» важнее, чем кажется.** Если там `None`, метод ничего не возвращает -
писать `lst = lst.append(40)` нельзя, список превратится в `None`.
