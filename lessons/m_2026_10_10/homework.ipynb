## Задача 1. Сумма до первого отрицательного

Напиши функцию `sum_until_negative`, которая принимает список чисел и возвращает
сумму чисел **до первого отрицательного**. Само отрицательное число и всё, что
идёт после него, в сумму не входит.

Примеры:

```
sum_until_negative([5, 3, -1, 10])  ->  8      # 5 + 3, дальше не считаем
sum_until_negative([1, 2])          ->  3      # отрицательных нет - сложили всё
sum_until_negative([-5, 1])         ->  0      # первое же число отрицательное
sum_until_negative([])              ->  0
```

````python
# ТВОЙ КОД
def sum_until_negative(numbers):
    total = 0
    for number in numbers:
        if number < 0:
            break
        total += number
    return total



print(sum_until_negative([5, 3, -1, 10]))
````

````python
# Проверка
assert sum_until_negative([5, 3, -1, 10]) == 8, \
    f"ожидалось 8, а вышло {sum_until_negative([5, 3, -1, 10])!r}"
assert sum_until_negative([1, 2]) == 3, \
    f"если отрицательных нет, нужно сложить всё: ожидалось 3, а вышло {sum_until_negative([1, 2])!r}"
assert sum_until_negative([-5, 1]) == 0, \
    f"первое число отрицательное - сумма 0, а вышло {sum_until_negative([-5, 1])!r}"
assert sum_until_negative([]) == 0, f"для пустого списка ожидается 0, а вышло {sum_until_negative([])!r}"
assert sum_until_negative([0, 4, -1]) == 4, (
    f"ожидалось 4, а вышло {sum_until_negative([0, 4, -1])!r}. "
    "Ноль не отрицательный - он должен попасть в сумму"
)
print("OK - задача 1 решена")
````

Исходный список менять нельзя.

## Задача 3. Все позиции буквы

Напиши функцию `positions`, которая принимает строку и букву и возвращает список
**всех индексов**, на которых эта буква встречается.

Метод `.index()` здесь не подойдёт: он находит только первое вхождение.

Примеры:

```
positions("барабан", "а")  ->  [1, 3, 5]
positions("ааа", "а")      ->  [0, 1, 2]
positions("кот", "я")      ->  []
```

````python
# ТВОЙ КОД
def positions(text, letter):
    indexes = []
    position = 0
    for char in text:
        if char == letter:
            indexes.append(position)
        position += 1
    return indexes

print(positions("барабан", "а"))
````
