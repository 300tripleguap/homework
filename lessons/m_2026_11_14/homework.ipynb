## Повторение: временная и пространственная сложность

Пиши в ячейках «ТВОЙ КОД», затем запускай проверку под задачей. После каждого решения коротко объясни его сложность в ячейке «ТВОЙ ОТВЕТ».

### Памятка

- Индекс и `len` у списка — O(1); `in`, `count`, `sum` — O(n).
- `append` у списка — O(1) амортизированно, то есть в среднем на операцию в длинной последовательности добавлений.
- Поиск и добавление в `set` / `dict` в этих задачах считаем O(1) в среднем. Построение множества из `n` элементов — O(n) в среднем.
- Последовательные действия складываются: O(n) + O(n) = O(n). Вложенные циклы оцениваем по числу повторений каждого.
- Срез списка создаёт копию. `sorted(items)` требует O(n log n) времени в худшем случае и O(n) памяти.
- Дополнительная память — память, созданная во время работы функции, без входных данных. **В этом ДЗ учитывай и возвращаемые списки и словари.**

Для оценок считаем, что одно число или короткая строка занимает O(1) памяти, а сравнение и арифметика с ними — O(1) времени. Указывай наиболее точную оценку роста. Время оценивай в худшем случае по расположению данных; для `set` и `dict` используй среднюю стоимость операций из памятки.

В задачах 2, 3 и 5 проверка замеряет время, в задаче 4 — память. Замеры помогают заметить медленное решение или лишний список, но сами по себе не доказывают O-оценку. Время зависит от компьютера; если проверка скорости не прошла, посмотри на алгоритм и повтори замер.

## Задача 5. Ускорь поиск первого повтора

Функция ниже возвращает число, которое **первым встретилось повторно при просмотре слева направо**. Если повторов нет, возвращает `None`. Вход — список целых чисел, менять его нельзя.

```python
first_repeat([5, 2, 2, 5])  # 2, а не 5: второй раз число 2 встретилось раньше
first_repeat([3, 1, 3, 1])  # 3
first_repeat([1, 2, 3])     # None
```

Заготовка даёт правильный ответ, но в худшем случае требует O(n^2) времени. **Перепиши её**, сохранив результат: нужно O(n) времени в среднем и не более O(n) дополнительной памяти.

**Подсказка, если застрял:** здесь важен порядок просмотра, но для хранения уже встреченных чисел порядок не нужен.


````python
# ТВОЙ КОД: ускорь функцию.
def first_repeat(numbers):
    seen = []
    for x in numbers:
        if x in seen:
            return x
        seen.append(x)
    return None
````

````python
# Проверка
import time

assert first_repeat([5, 2, 2, 5]) == 2, "Важен первый повтор при просмотре слева направо"
assert first_repeat([3, 1, 3, 1]) == 3
assert first_repeat([]) is None
assert first_repeat([1, 2, 3]) is None
assert first_repeat([8, 8, 8]) == 8
assert first_repeat([0, 1, 0]) == 0, "Ноль тоже может быть ответом"
assert first_repeat([-2, 4, -2]) == -2
data = [5, 2, 2, 5]
first_repeat(data)
assert data == [5, 2, 2, 5], "Не меняй исходный список"

# Без повторов приходится просмотреть весь вход.
big_unique = list(range(30_000))
start = time.perf_counter()
result_unique = first_repeat(big_unique)
result_repeat = first_repeat(big_unique + [17])
elapsed = time.perf_counter() - start
assert result_unique is None and result_repeat == 17, "Проверь длинный список"
assert elapsed < 1.5, f"Слишком медленно: {elapsed:.3f} с. Какую структуру ты используешь для поиска?"
print(f"Задача 5: верно, время двух вызовов {elapsed:.4f} с")
````
