# 1. Строки - `str`

## 1.1. Описание

**Строка - это текст:** последовательность символов в кавычках.

```python
name = "Егор"
```

Строкой хранят всё, что человек читает или вводит: имена, адреса, сообщения, номер телефона,
содержимое файлов. Всё, что вводит пользователь через `input()`, - тоже строка.

> **Почему номер телефона - строка, а не число?** У него бывают `+` и ведущие нули, а число их
> «съест»: `08123` превратится в `8123`. И арифметика с номером не нужна.
> Правило: **если с данными не нужна математика - это строка.**

Три свойства, которые нужно запомнить:

| Свойство | Что значит |
|---|---|
| Упорядоченность | У каждого символа есть номер (индекс) -> работают `s[0]` и срезы `s[1:4]` |
| Неизменяемость | Строку нельзя изменить «на месте». Методы **не меняют** строку, а **возвращают новую** |
| Любые символы | Кириллица, иероглифы, эмодзи - всё это валидные символы |

## 1.2. Примеры

### Пример 1. Создание строк, склейка, f-строки

````python
# Одинарные и двойные кавычки равноправны
first_name = "Егор"
last_name = 'Иванов'

print(len(first_name))               # 4 - количество символов
print(first_name + " " + last_name)  # склейка через +
print("-" * 30)                      # повторение через *

# f-строка - главный способ собрать текст: буква f перед кавычками,
# внутри {} можно подставлять переменные и даже вычисления
age = 15
print(f"Меня зовут {first_name} {last_name}, мне {age} лет.")
print(f"Через 5 лет мне будет {age + 5}")

# Кавычки внутри текста: берём строку в "другие" кавычки
print("Он сказал: 'поехали'")
````

### Пример 2. Индексы и срезы

Нумерация символов идёт **с нуля**, отрицательные индексы считают с конца:

```
 П   и   т   о   н
 0   1   2   3   4      <- слева направо
-5  -4  -3  -2  -1      <- справа налево
```

````python
word = "Питон"

print(word[0])    # П - первый символ
print(word[-1])   # н - последний (короче, чем word[len(word)-1])

# Срез [начало:конец] - конец НЕ включается
print(word[0:3])  # Пит  (символы 0, 1, 2)
print(word[:3])   # Пит  (начало можно опустить)
print(word[2:])   # тон  (конец можно опустить)

# Срез с шагом [начало:конец:шаг]
print(word[::-1])  # нотиП - шаг -1 переворачивает строку
````

### Пример 3. Методы и неизменяемость

Метод - действие, «прикреплённое» к объекту: пишется через точку - `строка.метод()`.

````python
print("привет".upper())            # ПРИВЕТ
print("егор иванов".title())       # Егор Иванов - каждое слово с заглавной
print("   егор   ".strip())        # егор - убрали пробелы по краям
print("2026-09-15".replace("-", "."))   # 2026.09.15
print("Иванов Егор Петрович".split())   # ['Иванов', 'Егор', 'Петрович']
print(", ".join(["хлеб", "молоко"]))    # хлеб, молоко
print("столица" in "Москва - столица")  # True

# Методы можно вызывать цепочкой - слева направо
print("   егор ИВАНОВ  ".strip().lower().title())  # Егор Иванов

# ГЛАВНАЯ ЛОВУШКА: строка неизменяема, метод её НЕ меняет
name = "егор"
name.upper()        # результат посчитан... и выброшен
print(name)         # егор - ничего не изменилось

name = name.upper() # правильно: сохраняем результат обратно в переменную
print(name)         # ЕГОР
````

# 2. Целые числа - `int`

## 2.1. Описание

**`int` - это целое число**, без дробной части: `0`, `7`, `-3`, `1000000`.

Им считают то, что не бывает «половинками»: количество товаров, возраст, год, номер квартиры,
индекс символа в строке.

Что важно знать:

| Факт | Пример |
|---|---|
| Размер не ограничен | `2 ** 200` посчитается целиком, без переполнения |
| Для читаемости можно ставить `_` | `1_000_000` - это то же самое, что `1000000` |
| Деление `/` всегда даёт `float` | `10 / 2` -> `5.0`, а не `5` |
| Целочисленное деление `//` и остаток `%` дают `int` | `10 // 3` -> `3`, `10 % 3` -> `1` |

## 2.2. Примеры

### Пример 1. Арифметика и два вида деления

````python
a = 10
b = 3

print(a + b)   # 13
print(a - b)   # 7
print(a * b)   # 30
print(a ** b)  # 1000 - возведение в степень

print(a / b)   # 3.3333333333333335 - обычное деление, тип float
print(a // b)  # 3   - целая часть от деления
print(a % b)   # 1   - остаток от деления

print(type(a / b))   # <class 'float'> - деление "портит" тип!
print(type(a // b))  # <class 'int'>
````

### Пример 2. Зачем нужны `//` и `%`

Эта пара - главный рабочий инструмент при работе с целыми числами.

````python
# Перевод единиц: 200 минут - это сколько часов и минут?
total_minutes = 200
print(total_minutes // 60, "ч", total_minutes % 60, "мин")  # 3 ч 20 мин

# Проверка на чётность: остаток от деления на 2
print(7 % 2)   # 1 - нечётное
print(8 % 2)   # 0 - чётное

# Последняя цифра числа - это остаток от деления на 10
print(472 % 10)   # 2
print(472 // 10)  # 47 - а так мы "отрезаем" последнюю цифру
````

### Пример 3. Превращение в число и обратно

````python
# Строка -> число
print(int("42") + 1)      # 43
# print(int("сорок два")) # ValueError: так нельзя

# Дробное -> целое: int() ОТБРАСЫВАЕТ дробную часть, а round() округляет
print(int(3.9))     # 3
print(round(3.9))   # 4
print(int(-3.9))    # -3

# Число -> строка (например, чтобы склеить с текстом)
count = 5
print("Товаров: " + str(count))
print(f"Товаров: {count}")   # то же самое, но короче

# Большие числа: разделители читаются человеком, но не влияют на значение
salary = 1_500_000
print(salary)   # 1500000
````

# 3. Дробные числа - `float`

## 3.1. Описание

**`float` - число с дробной частью:** `3.14`, `-0.5`, `99.99`. Разделитель - **точка**, не запятая.

Им измеряют то, что бывает «не целым»: рост, вес, температуру, проценты, среднее значение.

**Главное, что нужно знать про `float`:** он хранится в двоичной системе и поэтому **неточен**.
Компьютер не может записать `0.1` абсолютно точно - так же, как мы не можем записать 1/3
десятичной дробью (0.3333... никогда не кончится).

```python
0.1 + 0.2   # 0.30000000000000004
```

Для физики и статистики эта погрешность неважна. Для **денег** - критична: там используют
`Decimal` (следующий раздел).

## 3.2. Примеры

### Пример 1. Создание и арифметика

````python
height = 1.75
temperature = -2.5

print(type(height))        # <class 'float'>
print(height * 2)          # 3.5

# int + float = float: "заразный" тип, результат всегда дробный
print(2 + 0.5, type(2 + 0.5))   # 2.5 <class 'float'>

# Деление всегда даёт float, даже если делится нацело
print(10 / 5)   # 2.0

# Запись очень больших и очень маленьких чисел
print(1.5e3)    # 1500.0  (1.5 * 10 в степени 3)
print(2e-3)     # 0.002
````

### Пример 2. Неточность float и как с ней жить

````python
result = 0.1 + 0.2
print(result)             # 0.30000000000000004
print(result == 0.3)      # False - неожиданно, но так и есть

# Правило: НИКОГДА не сравнивайте float через == напрямую.
# Вариант 1 - округлить оба числа до нужной точности
print(round(result, 2) == 0.3)   # True

# Вариант 2 - math.isclose: "достаточно ли близки эти числа"
import math
print(math.isclose(result, 0.3))  # True

# Ещё пример той же природы
print(0.1 + 0.1 + 0.1)   # 0.30000000000000004
````

### Пример 3. Округление и красивый вывод

````python
pi = 3.14159

print(round(pi))        # 3   - без второго аргумента округляет до целого (тип int)
print(round(pi, 2))     # 3.14 - до двух знаков после точки

# Форматирование прямо в f-строке - самый частый способ вывести цену
price = 1234.5
print(f"{price:.2f}")        # 1234.50 - ровно 2 знака, с нулём на конце
print(f"{price:.0f}")        # 1234    - без дробной части
print(f"Цена: {price:.2f} руб.")

# Внимание: round() возвращает число, а f"{...:.2f}" - строку
print(type(round(price, 2)))   # <class 'float'>
print(type(f"{price:.2f}"))    # <class 'str'>
````

# 4. Точные дроби - `Decimal`

## 4.1. Описание

**`Decimal` - тип для точных десятичных вычислений.** Он считает так же, как человек на бумаге:
`0.1 + 0.2` даёт ровно `0.3`, без «хвоста».

Нужен везде, где ошибка в копейку недопустима: деньги, бухгалтерия, налоги, банковские операции.

Он не встроен в язык - его надо импортировать:

```python
from decimal import Decimal
```

**Железное правило:** создавать `Decimal` только **из строки** - `Decimal('0.1')`.
Если передать float (`Decimal(0.1)`), внутрь попадёт уже испорченное число, и весь смысл теряется.

## 4.2. Примеры

### Пример 1. Decimal против float

````python
from decimal import Decimal

print(0.1 + 0.2)                          # 0.30000000000000004 - float врёт
print(Decimal('0.1') + Decimal('0.2'))    # 0.3 - Decimal точен

print(0.1 + 0.2 == 0.3)                            # False
print(Decimal('0.1') + Decimal('0.2') == Decimal('0.3'))   # True

# Считаем чек: 3 товара по 19.99
print(0.1 * 3)                        # 0.30000000000000004
print(Decimal('19.99') * 3)           # 59.97 - ровно столько, сколько на ценнике
````

### Пример 2. Главная ловушка - создание из float

````python
from decimal import Decimal

# ПРАВИЛЬНО - из строки
print(Decimal('0.1'))   # 0.1

# НЕПРАВИЛЬНО - из float: испорченное число уже не починить
print(Decimal(0.1))     # 0.1000000000000000055511151231257827021181583404541015625

# Из целого числа можно смело - там нечему портиться
print(Decimal(100))     # 100
````

### Пример 3. Округление до копеек и смешивание типов

````python
from decimal import Decimal

price = Decimal('1234.56')
discount = Decimal('0.15')

raw_total = price * (1 - discount)
print(raw_total)   # 1049.3760 - лишние знаки, в кассе так нельзя

# quantize округляет до нужного числа знаков. Образец '0.01' = два знака.
total = raw_total.quantize(Decimal('0.01'))
print(total)       # 1049.38

# Decimal нельзя складывать с float - Python специально запрещает,
# чтобы неточность не просочилась в точные расчёты:
# print(Decimal('1.1') + 2.2)   # TypeError

# А с int - можно
print(Decimal('1.1') + 2)       # 3.1

# Обратно во float переводим только в самом конце, для вывода
print(float(total))             # 1049.38
````

Посчитай точную сумму чека из трёх товаров: `199.99`, `49.50` и `5.51`.

### Задача 2. Скидка с округлением до копеек

Цена `price = Decimal('2499.99')`, скидка 15%. Посчитай итог и округли его до копеек
через `.quantize()`. Ответ положи в `final_price` (ожидается `2124.99`).

````python
from decimal import Decimal

price = Decimal('2499.99')
discount = Decimal('0.15')

# ТВОЙ КОД
final_price = price * (1 - discount)
final_price = final_price.quantize(Decimal('0.01'))

print(final_price)
````

````python
# Проверка
assert isinstance(final_price, Decimal), "final_price должен быть Decimal"
assert final_price == Decimal('2124.99'), f"ожидалось 2124.99, а получилось {final_price}"
assert str(final_price) == "2124.99", f"нужно ровно два знака после точки, а получилось {final_price}"
print("OK - задача 2 решена")
````

# 5. Логический тип - `bool`

## 5.1. Описание

**`bool` - это ответ на вопрос «да или нет».** У него всего два значения: `True` и `False`
(с большой буквы, без кавычек).

Откуда он берётся:

- из сравнений: `5 > 3` -> `True`;
- из проверок: `"а" in "мама"` -> `True`;
- из методов: `"42".isdigit()` -> `True`.

Зачем нужен: на нём держится **любое условие** в программе - `if`, `while`, фильтры.
Пока условий мы не проходили, но именно `bool` решает, какая ветка кода выполнится.

**Операторов сравнения шесть:** `==` (равно), `!=` (не равно), `>`, `<`, `>=`, `<=`.
Не путай `=` (присвоить значение) и `==` (сравнить).

## 5.2. Примеры

### Пример 1. Сравнения

````python
age = 15

print(age > 18)    # False
print(age >= 15)   # True
print(age == 15)   # True  - сравнение
print(age != 15)   # False - "не равно"

print(type(age > 18))   # <class 'bool'>

# Сравнивать можно не только числа
print("abc" == "ABC")           # False - регистр важен
print("abc" == "ABC".lower())   # True
print(3 < 5 < 10)               # True - можно писать цепочкой, как в математике
````

### Пример 2. `and`, `or`, `not`

- `and` - истина, только если **оба** условия истинны;
- `or` - истина, если истинно **хотя бы одно**;
- `not` - переворачивает значение.

````python
age = 16
has_ticket = True
is_banned = False

print(age >= 16 and has_ticket)    # True  - оба условия выполнены
print(age >= 18 and has_ticket)    # False - первое не выполнено
print(age >= 18 or has_ticket)     # True  - хватило второго
print(not is_banned)               # True  - "не забанен"

# Всё вместе: пускать ли на сеанс
can_enter = age >= 16 and has_ticket and not is_banned
print(can_enter)   # True
````

### Пример 3. «Ложные» значения

В логическом смысле Python считает ложью не только `False`, но и всё «пустое»:
`0`, `""`, `[]`, `None`. Всё остальное - истина.

````python
print(bool(0))      # False
print(bool(-7))     # True - любое ненулевое число истинно
print(bool(""))     # False - пустая строка
print(bool(" "))    # True  - а вот пробел уже не пустота!
print(bool([]))     # False - пустой список
print(bool(None))   # False

# Забавный факт: bool - это разновидность int, поэтому его можно складывать
print(True + True)    # 2
print(True == 1)      # True
print(False == 0)     # True
````

# 6. Пустое значение - `None`

## 6.1. Описание

**`None` - это «ничего»:** значение отсутствует или пока неизвестно. Пишется с большой буквы,
без кавычек, и значение у этого типа ровно одно.

Зачем нужен отдельный тип для пустоты? Чтобы отличать **«нет данных»** от **«данные есть, и они нулевые»**:

| Значение | Что означает |
|---|---|
| `None` | Телефон не указан - мы его не знаем |
| `""` | Введена пустая строка |
| `0` | Скидка известна, и она равна нулю |

Разница важная: «скидки нет» и «про скидку ничего не известно» - совершенно разные ситуации.

**Правило:** `None` проверяют оператором `is`, а не `==`: `if phone is None`.

## 6.2. Примеры

### Пример 1. Что это такое

````python
phone = None

print(phone)         # None
print(type(phone))   # <class 'NoneType'>

# None - это НЕ строка "None" и не ноль
print(phone == "None")   # False
print(phone == 0)        # False
print(phone == "")       # False
````

### Пример 2. Проверка через `is`

````python
middle_name = None
nickname = ""

print(middle_name is None)      # True  - так правильно
print(nickname is None)         # False - пустая строка это не None
print(middle_name is not None)  # False - "значение задано?"

# Типичное применение: подставить заглушку, если данных нет
phone = None
phone_text = phone if phone is not None else "не указан"
print(f"Телефон: {phone_text}")
````

### Пример 3. Откуда `None` берётся сам собой

````python
# 1. Функция, которая ничего не возвращает, на самом деле возвращает None
result = print("привет")
print(result)   # None

# 2. Методы, которые меняют объект "на месте", тоже возвращают None.
#    Частая ошибка новичка:
numbers = [3, 1, 2]
numbers = numbers.sort()   # sort() сортирует список, но возвращает None
print(numbers)             # None - список потерян!

# Правильно так:
numbers = [3, 1, 2]
numbers.sort()             # без присваивания
print(numbers)             # [1, 2, 3]

# 3. В условии None ведёт себя как False
print(bool(None))   # False
````

# 7. Список - `list`

## 7.1. Описание

**Список - это несколько значений, хранящихся под одним именем**, в квадратных скобках через запятую:

```python
cars = ['bmw', 'audi', 'mercedes']
```

Зачем нужен: когда данных много и заводить отдельную переменную под каждое - бессмысленно
(список покупок, оценки за четверть, товары в корзине).

Список похож на строку - у него тоже есть индексы, срезы, `len()` и проверка `in`.
Но есть **ключевое отличие**:

| | Строка `str` | Список `list` |
|---|---|---|
| Изменяемость | **Нельзя** менять на месте | **Можно**: добавлять, удалять, заменять |
| Что хранит | Только символы | Что угодно, даже вперемешку: числа, строки, другие списки |

```python
s = "кот"
s[0] = "р"        # ошибка!

lst = ["кот"]
lst[0] = "рот"    # работает
```

## 7.2. Примеры

### Пример 1. Создание, индексы, срезы

````python
cars = ['bmw', 'audi', 'mercedes']

print(cars)          # ['bmw', 'audi', 'mercedes']
print(cars[0])       # bmw - первый элемент (нумерация с нуля)
print(cars[-1])      # mercedes - последний
print(len(cars))     # 3 - сколько элементов
print(cars[0:2])     # ['bmw', 'audi'] - срез, как у строк
print('audi' in cars)   # True

# В списке могут лежать разные типы
mixed = [1, 'текст', 3.5, True, None]
print(mixed)

# Пустой список - нормальное начало, потом наполним
basket = []
print(len(basket))   # 0
````

### Пример 2. Изменение списка

````python
cars = ['bmw', 'audi', 'mercedes']

cars.append('toyota')      # добавить в конец
print(cars)                # ['bmw', 'audi', 'mercedes', 'toyota']

cars.insert(1, 'priora')   # вставить по индексу, остальные сдвинутся вправо
print(cars)                # ['bmw', 'priora', 'audi', 'mercedes', 'toyota']

cars[0] = 'volvo'          # заменить элемент по индексу
print(cars)                # ['volvo', 'priora', ...]

del cars[0]                # удалить по индексу
cars.remove('priora')      # удалить по значению
last = cars.pop()          # удалить последний И вернуть его
print(cars, "| достали:", last)
````

# 8. Время - `timestamp` и `datetime`

## 8.1. Описание

**timestamp - это момент времени, записанный одним числом:** сколько секунд прошло с
00:00 1 января 1970 года (эта дата называется «эпохой Unix»).

По типу это обычный `float`, например `1789123456.78`.

Зачем так странно? Число удобно хранить в базе, сравнивать (`t1 < t2` - что было раньше)
и вычитать (разница в секундах). А вот читать его человеку невозможно - поэтому для вывода
timestamp превращают в `datetime`.

| Что | Чем пользуемся |
|---|---|
| Текущий timestamp | `time.time()` |
| timestamp -> дата | `datetime.fromtimestamp(ts)` |
| Текущая дата и время сразу | `datetime.now()` |
| datetime -> красивая строка | `.strftime("%d.%m.%Y")` |
| Строка -> datetime | `datetime.strptime(text, "%d.%m.%Y")` |

Коды формата: `%d` - день, `%m` - месяц, `%Y` - год (4 цифры), `%H` - часы, `%M` - минуты, `%S` - секунды.

## 8.2. Примеры

### Пример 1. Получаем время и делаем его читаемым

````python
import time
from datetime import datetime

ts = time.time()
print(ts)          # 1789123456.789 - число секунд с 1970 года
print(type(ts))    # <class 'float'>

# Превращаем в объект даты
dt = datetime.fromtimestamp(ts)
print(dt)          # 2026-09-15 14:30:56.789

# И в читаемую строку
print(dt.strftime("%d.%m.%Y %H:%M"))   # 15.09.2026 14:30

# Короткий путь, если timestamp не нужен
print(datetime.now().strftime("%d.%m.%Y"))
````

### Пример 2. Из даты в строку и обратно

````python
from datetime import datetime

# Создаём конкретную дату: год, месяц, день
birthday = datetime(2011, 3, 25)
print(birthday.strftime("%d.%m.%Y"))   # 25.03.2011

# Разбираем строку в дату - шаблон должен ТОЧНО совпадать с текстом
text = "15.09.2026 18:30"
parsed = datetime.strptime(text, "%d.%m.%Y %H:%M")
print(parsed)          # 2026-09-15 18:30:00
print(parsed.year, parsed.month, parsed.day)   # 2026 9 15

# Обратно в timestamp
print(parsed.timestamp())
````

### Пример 3. Разница между датами

````python
from datetime import datetime, timedelta

start = datetime(2026, 9, 1)
end = datetime(2026, 9, 15)

delta = end - start         # разность дат даёт объект timedelta
print(delta)                # 14 days, 0:00:00
print(delta.days)           # 14 - целых дней
print(delta.total_seconds())  # 1209600.0 - всего секунд

# К дате можно прибавлять промежутки
print(start + timedelta(days=30))    # 2026-10-01 00:00:00
print(start + timedelta(hours=12))   # 2026-09-01 12:00:00
````
