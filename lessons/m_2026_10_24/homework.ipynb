## Задача 5. Ускорь функцию

Функция ниже убирает повторы, сохраняя порядок первого появления. Результат она даёт **правильный**, но работает за O(n^2): `x not in result` каждый раз пробегает весь накопленный список.

Перепиши её так, чтобы она работала за O(n) и давала тот же результат. Подсказка: нужна вторая структура данных, в которой проверка наличия стоит O(1).

````python
# ТВОЙ КОД - ускорь функцию, результат менять нельзя
def unique_in_order(items):
    result = []
    for x in items:
        if x not in result:
            result.append(x)
    return result


print(unique_in_order([3, 1, 3, 2, 1]))
````

````python
# Проверка
import time

assert unique_in_order([3, 1, 3, 2, 1]) == [3, 1, 2], "Порядок первого появления должен сохраняться"
assert unique_in_order([]) == [], "Пустой список"
assert unique_in_order([7, 7, 7]) == [7], "Неверный результат на [7, 7, 7]"
assert unique_in_order(["б", "а", "б"]) == ["б", "а"], "Должно работать и со строками"

big = list(range(20_000)) * 2

start = time.perf_counter()
result = unique_in_order(big)
elapsed = time.perf_counter() - start

assert result == list(range(20_000)), "Неверный результат на большом списке"
assert elapsed < 1.0, f"Слишком медленно: {elapsed:.2f} с. Проверка наличия всё ещё идёт по списку"

print(f"Задача 5: верно, время {elapsed:.4f} с")
````
