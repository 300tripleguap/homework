## 1. Зачем нужны функции

Посмотри на код ниже. Одно и то же действие - цена плюс НДС - записано три раза.

Что с ним не так:

- если ставка НДС изменится, править придётся в трёх местах, и одно легко забыть;
- ошибка в формуле размножается втрое;
- читая код, каждый раз приходится разбирать формулу заново.

**Функция - это имя, данное куску кода.** Написал один раз, дальше просто вызываешь по имени.

````python
# Было: одно и то же три раза
price1 = 1000
total1 = price1 * 1.2

price2 = 2500
total2 = price2 * 1.2

price3 = 700
total3 = price3 * 1.2

print(total1, total2, total3)

# Стало: формула в одном месте
def with_vat(price):
    return price * 1.2

print(with_vat(1000), with_vat(2500), with_vat(700))
````

## 2. Как объявить и вызвать функцию

```python
def имя_функции(параметры):
    тело функции
    return результат
```

Разберём по частям:

| Часть | Зачем |
|---|---|
| `def` | Ключевое слово: «объявляю функцию» |
| `имя_функции` | Как её потом вызывать. Имена пишут строчными, слова через `_` |
| `(параметры)` | Что функция принимает снаружи. Скобки обязательны, даже если пусто |
| `:` | Двоеточие в конце строки. Без него - `SyntaxError` |
| Тело | Код функции, с **отступом в 4 пробела** |
| `return` | Что функция отдаёт наружу |

**Два разных момента, которые часто путают:**

- **объявление** (`def ...`) - Python только запоминает функцию, код внутри не выполняется;
- **вызов** (`имя_функции()`) - вот теперь код выполняется.

````python
def greet(name):
    print("Python дошёл до тела функции")
    return f"Привет, {name}!"

print("1. Функция объявлена, но тело ещё не выполнялось")

message = greet("Егор")     # вот сейчас выполнится
print("2.", message)

print("3.", greet("Аня"))   # можно вызывать сколько угодно раз

# Функция без параметров - скобки всё равно нужны
def say_hi():
    return "Привет!"

print(say_hi())     # Привет!
print(say_hi)       # <function say_hi ...> - без скобок это сама функция, а не результат
````

## 3. Параметры и аргументы

- **Параметр** - имя в объявлении функции: `def power(base, exp)`.
- **Аргумент** - конкретное значение при вызове: `power(2, 3)`.

Аргументы можно передавать двумя способами:

| Способ | Пример | Когда удобно |
|---|---|---|
| Позиционно | `power(2, 3)` | Аргументов мало и порядок очевиден |
| По имени | `power(exp=3, base=2)` | Аргументов много; порядок тогда не важен |

**Значение по умолчанию** - параметр, который можно не передавать:

```python
def with_vat(price, vat=0.2):
    ...
```

Правило: параметры со значением по умолчанию пишутся **после** обычных.
`def f(a=1, b)` - синтаксическая ошибка.

````python
def with_vat(price, vat=0.2):
    return price * (1 + vat)

print(with_vat(1000))            # 1200.0 - vat взят по умолчанию
print(with_vat(1000, 0.1))       # 1100.0 - позиционно
print(with_vat(1000, vat=0.1))   # 1100.0 - по имени
print(with_vat(vat=0.1, price=1000))  # 1100.0 - по имени порядок не важен

# Частая ошибка: перепутан порядок позиционных аргументов
def divide(a, b):
    return a / b

print(divide(10, 2))   # 5.0
print(divide(2, 10))   # 0.2 - Python не догадается, что ты имел в виду другое
````

## 4. `return` против `print` - главное в теме

Это разные вещи, и путаница между ними - ошибка номер один у новичков.

| | `print(x)` | `return x` |
|---|---|---|
| Что делает | Показывает значение на экране | **Отдаёт** значение наружу |
| Можно сохранить в переменную? | Нет | Да |
| Можно использовать в вычислении? | Нет | Да |
| Видно ли в консоли? | Да | Нет (пока сам не напечатаешь) |

Функция с `print` вместо `return` выглядит рабочей, но её результат нельзя использовать дальше:
складывать, сравнивать, класть в список.

**Если `return` в функции нет, она возвращает `None`.**

````python
def area_print(a, b):
    print(a * b)          # только показали

def area_return(a, b):
    return a * b          # отдали наружу

area_print(3, 4)          # 12 напечаталось
x = area_print(3, 4)      # 12 снова напечаталось...
print("x =", x)           # ...но x = None, значения нет

y = area_return(3, 4)     # ничего не напечаталось
print("y =", y)           # y = 12 - значение сохранилось

print(area_return(3, 4) + area_return(2, 5))   # 22 - с результатом можно считать
# print(area_print(3, 4) + 1)   # TypeError: None + 1
````

### `return` завершает функцию

Как только Python доходит до `return`, функция заканчивается. Код после него не выполнится.

Поэтому `return` часто ставят в разных ветках `if` - какая ветка сработала, то и вернулось.

````python
def check(age):
    if age < 0:
        return "Неверный возраст"      # выходим сразу
    if age < 18:
        return "Несовершеннолетний"
    return "Совершеннолетний"          # сюда дойдём, только если оба if не сработали

print(check(-5))
print(check(10))
print(check(30))

def broken():
    return "первый"
    print("эта строка недостижима")    # выполнена не будет

print(broken())
````

## 5. Где живут переменные: локальная область

Переменная, созданная внутри функции, существует **только внутри неё**.
Снаружи её нет - Python скажет `NameError`.

Это не ограничение, а удобство: можно смело называть переменные внутри функций
`total`, `result`, `i`, не боясь испортить что-то в остальной программе.

````python
def calc():
    result = 42          # локальная переменная
    return result

print(calc())            # 42
# print(result)          # NameError: result не существует снаружи

# У каждого вызова свои переменные - они не мешают друг другу
def counter(start):
    n = start + 1
    return n

print(counter(10), counter(100))   # 11 101
````

## 6. LEGB - как Python ищет переменную

Когда в коде встречается имя, Python ищет его **по очереди** в четырёх местах
и останавливается на первом найденном. Порядок и называется **LEGB**:

| Буква | Область | Что это | Пример |
|---|---|---|---|
| **L** | Local | Внутри текущей функции | `x` создан прямо в этой функции |
| **E** | Enclosing | Во внешней функции (если эта вложена в другую) | `x` создан в функции, внутри которой мы находимся |
| **G** | Global | На верхнем уровне файла | `x` создан вне всех функций |
| **B** | Built-in | Встроенные имена Python | `len`, `sum`, `print`, `max` |

Нашёл на первом шаге - дальше не ищет. Не нашёл нигде - `NameError`.

````python
x = "глобальная (G)"

def outer():
    x = "внешняя (E)"

    def inner():
        x = "локальная (L)"
        print("inner видит:", x)     # L - своя есть, дальше не смотрит

    inner()
    print("outer видит:", x)         # своя локальная для outer

outer()
print("снаружи видно:", x)           # G - глобальная не изменилась
````

### Что будет, если своей переменной нет

Python пойдёт наружу по цепочке L -> E -> G -> B.

````python
x = "глобальная (G)"

def outer():
    x = "внешняя (E)"

    def inner():
        print("inner нашёл:", x)     # своей нет -> берёт из outer (E)

    inner()

outer()

def lonely():
    print("lonely нашёл:", x)        # ни своей, ни внешней функции -> берёт G

lonely()
````

### Важно: присваивание создаёт **локальную** переменную

Если внутри функции написать `x = ...`, Python создаст **новую локальную** переменную,
а глобальная останется прежней. Функция не может случайно испортить внешние данные -
и это хорошо.

````python
count = 10

def try_change():
    count = 999          # создалась НОВАЯ локальная переменная
    print("внутри:", count)

try_change()
print("снаружи:", count)     # 10 - глобальная не тронута
````

### Буква B: встроенные имена

`len`, `sum`, `max`, `min`, `print`, `int`, `str` - это не ключевые слова, а обычные имена
из встроенной области. Их **можно случайно перекрыть** своей переменной, и функция перестанет работать.

Правило: не называй переменные `len`, `sum`, `list`, `str`, `max`, `type`, `id`.

````python
print(len("Привет"))     # 6 - len найден в B

len = 5                  # создали свою переменную с тем же именем (в G)
print("теперь len =", len)
# print(len("Привет"))   # TypeError: 'int' object is not callable

del len                  # удалили свою - встроенная снова видна
print(len("Привет"))     # 6
````

## 7. Частые ошибки

| Ошибка | Как выглядит | Как правильно |
|---|---|---|
| `print` вместо `return` | `def f(): print(1)` -> результат `None` | `def f(): return 1` |
| Забыли вызвать | `print(say_hi)` -> `<function ...>` | `print(say_hi())` |
| Нет двоеточия | `def f(x)` -> `SyntaxError` | `def f(x):` |
| Нет отступа в теле | `SyntaxError: expected an indented block` | 4 пробела перед каждой строкой тела |
| Код после `return` | Не выполняется никогда | Перенести до `return` |
| Значение по умолчанию раньше обычного | `def f(a=1, b)` -> `SyntaxError` | `def f(b, a=1)` |
| Переменная функции снаружи | `NameError` | Вернуть её через `return` |
| Имя перекрыло встроенное | `len = 5`, дальше `len(...)` падает | Назвать иначе: `length`, `total` |

## Памятка

```python
def имя(обязательный, необязательный=значение):
    # тело с отступом 4 пробела
    return результат          # без return функция вернёт None

результат = имя(1)            # позиционно
результат = имя(1, необязательный=2)   # по имени
```

**LEGB:** Local -> Enclosing -> Global -> Built-in. Ищет в этом порядке, берёт первое найденное.
