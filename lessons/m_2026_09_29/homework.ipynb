## Задача 2. Цена с НДС

Напиши функцию `with_vat`, которая принимает цену и ставку НДС и возвращает итоговую цену,
округлённую до двух знаков.

Ставка по умолчанию - `0.2`, то есть вызов `with_vat(1000)` должен работать без второго аргумента.

Примеры:

```
with_vat(1000)        ->  1200.0
with_vat(1000, 0.1)   ->  1100.0
with_vat(999.99)      ->  1199.99
```

````python
print
````

````python
# ТВОЙ КОД
def with_vat(price, vat=0.2): 
    """
        price - позиционный параметр (без значения по умолчанию, он обязательный), 
        vat - параметр со значением по умолчанию, не обязательный
    """
    result = round(price * (vat + 1), 2)
    return result

print(
    # with_vat(1000), 
    with_vat(1000, 0.2), 
   # with_vat(999.99)
)
````

````python
# Проверка
assert with_vat(1000) == 1200.0, f"with_vat(1000) должно быть 1200.0, а вышло {with_vat(1000)}"
assert with_vat(1000, 0.1) == 1100.0, f"with_vat(1000, 0.1) должно быть 1100.0, а вышло {with_vat(1000, 0.1)}"
assert with_vat(999.99) == 1199.99, (
    f"with_vat(999.99) должно быть 1199.99, а вышло {with_vat(999.99)}. "
    "Не забудь округлить результат до двух знаков"
)
# assert with_vat(0.5, 500) == 750.0, "функция должна работать и при передаче аргументов по имени"
assert with_vat(vat=0.5, price=500) == 750.0, "функция должна работать и при передаче аргументов по имени"
print("OK - задача 2 решена")
````

## Задача 3. Отчёт по числам

Напиши функцию `stats_text`, которая принимает список чисел и возвращает строку:

```
сумма: 19, максимум: 5, среднее: 3.8
```

Среднее округли до двух знаков. Циклы не нужны - хватит готовых функций `sum`, `max` и `len`.

Примеры:

```
stats_text([4, 5, 3, 5, 2])  ->  "сумма: 19, максимум: 5, среднее: 3.8"
stats_text([1, 2])           ->  "сумма: 3, максимум: 2, среднее: 1.5"
```

````python
# ТВОЙ КОД
def stats_text(numbers):
    result = f"сумма: {sum(numbers)}, максимум: {max(numbers)}, среднее: {round(sum(numbers)/len(numbers),2)}"
    return result


print(stats_text([4, 5, 3, 5, 2]))
````

````python
# Проверка
assert stats_text([4, 5, 3, 5, 2]) == "сумма: 19, максимум: 5, среднее: 3.8", \
    f"получилось {stats_text([4, 5, 3, 5, 2])!r}"
assert stats_text([1, 2]) == "сумма: 3, максимум: 2, среднее: 1.5", \
    f"на другом списке сломалось: {stats_text([1, 2])!r}"
assert stats_text([10]) == "сумма: 10, максимум: 10, среднее: 10.0", \
    f"на списке из одного числа сломалось: {stats_text([10])!r}"
assert stats_text([1, 2, 2]) == "сумма: 5, максимум: 2, среднее: 1.67", (
    f"получилось {stats_text([1, 2, 2])!r}. "
    "Среднее 5/3 нужно округлить до двух знаков"
)
print("OK - задача 3 решена")
````
