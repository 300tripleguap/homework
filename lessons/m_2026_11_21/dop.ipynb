# Списковое включение

```python
items = [1, 2, 3]
```

```python
positive = []
for x in items:
    if x > 0:
        positive.append(x)
```

```python
# тоже самое что и в прошлой ячейке, только это называется списковое включение
positive = [x for x in items if x > 0]
```

# Декоратор

```python
def fun():
    print("Выполнение работы")
```

```python
fun()
```

```python
def decorator(fun):
    def wrapper():
        print("Начало")  # декорирование, т.е. добавление нового функционала к fun
        fun()  # fun сохраняется в wrapper и это называется замыкание
        print("Конец")  # декорирование
    return wrapper
```

```python
def fun():
    print("Выполнение работы")
```

```python
fun = decorator(fun)
fun()
```

```python
@decorator
def fun():
    print("Выполнение работы")

fun()
```

```python
import time

# Декоратор таймер
def timer(fun):
    def wrapper():
        start_time = time.time()
        fun()
        print("Время выполнения: ", time.time() - start_time)
    return wrapper
```

```python
@timer
def fun():
    print("Выполнение работы")
    time.sleep(3)

fun()
```

```python
@timer
def fun2():
    print("Выполнение работы")
    time.sleep(2)

fun2()
```
