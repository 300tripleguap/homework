## Задача 1. Цена билета

Напиши функцию `ticket_price`, которая принимает возраст и возвращает цену билета:

| Возраст | Цена |
|---|---|
| меньше 0 | `"Неверный возраст"` |
| от 0 до 6 включительно | `0` |
| от 7 до 17 включительно | `100` |
| от 18 до 64 включительно | `300` |
| 65 и больше | `150` |

Проверка пройдётся ровно по границам диапазонов - следи, где нужен `<`, а где `<=`.

````python
# ТВОЙ КОД
def ticket_price(age):
    if age < 0:
        return ('Неверный возраст')
    elif age <= 6:
        return 0
    elif age <= 17:
        return 100
    elif age <= 64:
        return 300
    elif age >= 65:
        return 150
    
print(ticket_price(-1), ticket_price(6), ticket_price(7), ticket_price(65))
````

````python
# Проверка - границы диапазонов
assert ticket_price(-1) == "Неверный возраст", f"для -1 ожидалось 'Неверный возраст', а вышло {ticket_price(-1)!r}"
assert ticket_price(0) == 0, f"для 0 ожидалось 0, а вышло {ticket_price(0)!r}"
assert ticket_price(6) == 0, f"для 6 ожидалось 0, а вышло {ticket_price(6)!r}"
assert ticket_price(7) == 100, f"для 7 ожидалось 100, а вышло {ticket_price(7)!r}"
assert ticket_price(17) == 100, f"для 17 ожидалось 100, а вышло {ticket_price(17)!r}"
assert ticket_price(18) == 300, f"для 18 ожидалось 300, а вышло {ticket_price(18)!r}"
assert ticket_price(64) == 300, f"для 64 ожидалось 300, а вышло {ticket_price(64)!r}"
assert ticket_price(65) == 150, f"для 65 ожидалось 150, а вышло {ticket_price(65)!r}"
assert ticket_price(90) == 150, f"для 90 ожидалось 150, а вышло {ticket_price(90)!r}"
print("OK - задача 1 решена")
````

Напиши функцию `can_enter`, которая принимает три значения `True` / `False`
и возвращает `True` или `False` по правилу:

> Пускают того, кто **участник или тренер**, и **при этом** у него есть пропуск.

Проверка переберёт **все восемь** возможных комбинаций - формула должна быть верной
не для одного случая, а для всех.

````python
# ТВОЙ КОД
def can_enter(is_member, is_coach, has_pass):
    return (is_member or is_coach) and has_pass
        


print(can_enter(True, False, True))
````

````python
# Проверка - все восемь комбинаций
#                участник, тренер, пропуск     ожидаем
assert can_enter(False, False, False) is False, "никто и без пропуска - не пускаем"
assert can_enter(False, False, True)  is False, "пропуск есть, но человек ни участник, ни тренер"
assert can_enter(False, True,  False) is False, "тренер, но БЕЗ пропуска - не пускаем"
assert can_enter(False, True,  True)  is True,  "тренер с пропуском - пускаем"
assert can_enter(True,  False, False) is False, "участник, но без пропуска - не пускаем"
assert can_enter(True,  False, True)  is True,  "участник с пропуском - пускаем"
assert can_enter(True,  True,  False) is False, "и участник, и тренер, но БЕЗ пропуска - не пускаем"
assert can_enter(True,  True,  True)  is True,  "всё есть - пускаем"
print("OK - задача 2 решена")
````

## Задача 3. Считаем положительные

Напиши функцию `count_positive`, которая принимает список чисел и возвращает,
**сколько** в нём чисел больше нуля.

Примеры:

```
count_positive([1, -2, 0, 5])  ->  2
count_positive([-1, -2])       ->  0
count_positive([])             ->  0
```

````python
# ТВОЙ КОД
def count_positive(numbers):
    count = 0
    for i in numbers:
        if i > 0:
            count += 1
    return count


print(count_positive([1, -2, 0, 5]))
````

````python
# Проверка
assert count_positive([1, -2, 0, 5]) == 2, f"ожидалось 2, а вышло {count_positive([1, -2, 0, 5])!r}"
assert count_positive([-1, -2]) == 0, f"отрицательных считать не нужно, а вышло {count_positive([-1, -2])!r}"
assert count_positive([]) == 0, f"для пустого списка ожидается 0, а вышло {count_positive([])!r}"
assert count_positive([3, 3, 3]) == 3, f"ожидалось 3, а вышло {count_positive([3, 3, 3])!r}"
assert count_positive([0, 0]) == 0, "ноль не больше нуля - его считать не нужно"
print("OK - задача 3 решена")
````

## Задача 4. Сумма чётных

Напиши функцию `even_sum`, которая принимает число `n` и возвращает сумму всех
**чётных** чисел от 1 до `n` включительно.

Примеры:

```
even_sum(10)  ->  30      # 2 + 4 + 6 + 8 + 10
even_sum(5)   ->  6       # 2 + 4
even_sum(1)   ->  0       # чётных нет
```

Следи за правой границей `range` - она не включается.

````python
# ТВОЙ КОД
def even_sum(n):
    total = 0
    for summa in range(1, n + 1):
        if summa % 2 == 0:    
            total += summa
    return total    



print(even_sum(10))
````

````python
# Проверка
assert even_sum(10) == 30, f"для 10 ожидалось 30, а вышло {even_sum(10)!r}"
assert even_sum(5) == 6, f"для 5 ожидалось 6, а вышло {even_sum(5)!r}"
assert even_sum(1) == 0, f"для 1 ожидалось 0, а вышло {even_sum(1)!r}"
assert even_sum(2) == 2, (
    f"для 2 ожидалось 2, а вышло {even_sum(2)!r}. "
    "Само число n тоже должно попадать в перебор"
)
assert even_sum(100) == 2550, f"для 100 ожидалось 2550, а вышло {even_sum(100)!r}"
print("OK - задача 4 решена")
````

## Задача 5. Сколько раз удвоить

Напиши функцию `double_until`, которая принимает начальное число `start` и порог `limit`
и возвращает, **сколько раз нужно удвоить** `start`, чтобы он стал не меньше `limit`.

Здесь заранее неизвестно, сколько понадобится шагов, - значит, нужен `while`.

Примеры:

```
double_until(1, 10)   ->  4     # 2, 4, 8, 16
double_until(3, 20)   ->  3     # 6, 12, 24
double_until(5, 5)    ->  0     # уже не меньше порога, удваивать не надо
```

````python
# ТВОЙ КОД
def double_until(start, limit):
    count = 0
    while start < limit:
        start *= 2
        count += 1
    return count
print(double_until(1, 10))
````

````python
# Проверка
assert double_until(1, 10) == 4, f"ожидалось 4, а вышло {double_until(1, 10)!r}"
assert double_until(3, 20) == 3, f"ожидалось 3, а вышло {double_until(3, 20)!r}"
assert double_until(5, 5) == 0, (
    f"ожидалось 0, а вышло {double_until(5, 5)!r}. "
    "Если число уже достигло порога, цикл не должен выполниться ни разу"
)
assert double_until(2, 100) == 6, f"ожидалось 6, а вышло {double_until(2, 100)!r}"
assert double_until(10, 3) == 0, "если start уже больше limit, ответ 0"
print("OK - задача 5 решена")
````
