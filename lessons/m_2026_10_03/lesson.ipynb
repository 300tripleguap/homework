## 1. Зачем нужны циклы

На прошлых занятиях, чтобы вывести список, приходилось писать по строке на каждый элемент.
Это работает ровно до тех пор, пока элементов три. А если их триста? А если заранее
неизвестно, сколько их будет?

**Цикл - это «повтори это действие для каждого элемента»**. Код пишется один раз,
а выполняется столько раз, сколько нужно.

````python
marks = [5, 3, 4]

# Было: по строке на элемент
print(marks[0])
print(marks[1])
print(marks[2])

print("---")

# Стало: одна запись на любую длину списка
for mark in marks:
    print(mark)
````

## 2. `for` - перебор коллекции

```python
for переменная in коллекция:
    тело цикла
```

Как это читается: «для каждого элемента из коллекции - выполни тело».

| Часть | Что означает |
|---|---|
| `for` | Ключевое слово |
| `переменная` | Имя, которое ты придумываешь сам. На каждом шаге в неё кладётся очередной элемент |
| `in коллекция` | Что перебираем: список, строку, `range` |
| `:` | Двоеточие обязательно |
| Тело | Код с **отступом в 4 пробела**. Всё, что с отступом, повторяется |

Переменную цикла называют по смыслу: `for mark in marks`, `for word in words`.
Так код читается как обычный текст.

Строку тоже можно перебирать - по символам.

````python
words = ["дом", "солнце", "кот"]

for word in words:
    print(word, "-", len(word), "букв")

print("---")

# Строка перебирается посимвольно
for letter in "Пайтон":
    print(letter, end=" ")
print()

print("---")

# Тело может быть из нескольких строк - главное, отступ
for word in words:
    big = word.upper()
    print(big)

# А эта строка уже БЕЗ отступа - она вне цикла и выполнится один раз
print("Цикл закончился")
````

## 3. `range` - перебор чисел

Когда перебирать нужно не элементы списка, а числа, используют `range`.

| Запись | Что даёт |
|---|---|
| `range(5)` | `0, 1, 2, 3, 4` - пять чисел, начиная с нуля |
| `range(1, 5)` | `1, 2, 3, 4` - от 1 до 5 |
| `range(1, 10, 2)` | `1, 3, 5, 7, 9` - от 1 до 10 с шагом 2 |
| `range(10, 0, -1)` | `10, 9, 8, ... 1` - обратный отсчёт |

**Главное правило: правая граница НЕ включается.** `range(1, 5)` даёт числа до 4.
Чтобы дойти до 5, пишут `range(1, 6)`.

`range` не создаёт список - он выдаёт числа по одному, по мере надобности.
Поэтому `print(range(5))` покажет не числа, а `range(0, 5)`. Чтобы увидеть числа,
оберни в `list(...)`.

````python
for i in range(5):
    print(i, end=" ")
print()

for i in range(1, 6):        # 1..5, шестёрка не войдёт
    print(i, end=" ")
print()

for i in range(0, 11, 2):    # чётные от 0 до 10
    print(i, end=" ")
print()

for i in range(3, 0, -1):    # обратный отсчёт
    print(i, end=" ")
print("Поехали!")

print(range(5))              # range(0, 5) - это не список
print(list(range(5)))        # [0, 1, 2, 3, 4] - а вот теперь числа видно

# range по длине списка: так можно получить и индекс, и элемент
cars = ["bmw", "audi", "kia"]
for i in range(len(cars)):
    print(i, cars[i])
````

## 4. Накопление результата

Сам по себе цикл ничего не запоминает: на каждом шаге переменная цикла просто меняется.
Чтобы получить результат, заводят переменную **до** цикла и меняют её **внутри**.

Три типовых приёма - их стоит запомнить как шаблоны:

| Задача | Заготовка до цикла | Действие внутри |
|---|---|---|
| Сумма | `total = 0` | `total = total + x` (короче: `total += x`) |
| Счётчик | `count = 0` | `count += 1` |
| Новый список | `result = []` | `result.append(x)` |

Заводить переменную **до** цикла обязательно: если написать `total = 0` внутри,
она будет обнуляться на каждом шаге.

````python
prices = [100, 250, 50, 300]

# Сумма
total = 0
for price in prices:
    total += price          # то же самое, что total = total + price
print("Сумма:", total)

# Счётчик с условием
count = 0
for price in prices:
    if price >= 100:
        count += 1
print("Дорогих товаров:", count)

# Новый список
with_discount = []
for price in prices:
    with_discount.append(price * 0.9)
print("Со скидкой:", with_discount)

# Поиск максимума вручную (хотя для списка проще max())
biggest = prices[0]
for price in prices:
    if price > biggest:
        biggest = price
print("Максимум:", biggest)
````

## 5. `while` - цикл с условием

`for` используют, когда известно, **что** перебирать. `while` - когда известно только
**условие продолжения**, а число повторений заранее неизвестно.

```python
while условие:
    тело цикла
```

Читается: «пока условие истинно - повторяй тело».

**Главная опасность - бесконечный цикл.** Если внутри тела ничего не меняет условие,
программа зависнет. В блокноте такую ячейку придётся останавливать кнопкой «стоп».

Поэтому в `while` почти всегда есть три части:

1. переменная заводится **до** цикла;
2. условие её проверяет;
3. тело её **меняет** - иначе цикл никогда не кончится.

````python
# 1. Счётчик до цикла
n = 1
# 2. Условие
while n <= 5:
    print(n, end=" ")
    # 3. Изменение - без этой строки цикл был бы бесконечным
    n += 1
print()

# Пример, где for неудобен: неизвестно, сколько шагов понадобится
money = 1000
years = 0
while money < 2000:
    money = money * 1.1      # +10% в год
    years += 1
print(f"Удвоение через {years} лет, итог: {money:.2f}")

# Обратный отсчёт
count = 3
while count > 0:
    print(count)
    count -= 1
print("Старт!")
````

## 6. `break` и `continue`

Два способа управлять ходом цикла изнутри:

| Команда | Что делает |
|---|---|
| `break` | **Прерывает весь цикл**. Дальше выполняется код после цикла |
| `continue` | **Пропускает текущий шаг** и переходит к следующему |

`break` нужен, когда нашёл искомое и дальше перебирать бессмысленно.
`continue` - когда часть элементов нужно просто пропустить.

````python
numbers = [3, 8, 12, 5, 20]

# break: нашли первое число больше 10 и вышли
for n in numbers:
    if n > 10:
        print("Нашли:", n)
        break               # 5 и 20 уже не проверяются
    print("проверили", n)

print("---")

# continue: пропускаем короткие слова
words = ["дом", "солнце", "кот", "машина"]
long_words = []
for word in words:
    if len(word) <= 3:
        continue            # остальное тело на этом шаге не выполнится
    long_words.append(word)
print(long_words)

print("---")

# break в while - частый способ выйти из цикла по событию
attempts = [10, 25, 42, 7]
i = 0
while i < len(attempts):
    if attempts[i] == 42:
        print("Угадал на попытке", i + 1)
        break
    i += 1
````

## 7. `enumerate` - индекс вместе со значением

Иногда в цикле нужен не только элемент, но и его номер: пронумеровать список,
вывести позицию найденного.

Можно через `range(len(...))`, но есть способ короче и понятнее:

```python
for индекс, значение in enumerate(коллекция):
```

На каждом шаге `enumerate` отдаёт **пару**: номер и элемент. Нумерация начинается с нуля,
но её можно сдвинуть: `enumerate(коллекция, start=1)`.

````python
cars = ["bmw", "audi", "kia"]

# Было: через range(len(...))
for i in range(len(cars)):
    print(i, cars[i])

print("---")

# Стало: через enumerate - короче и читается лучше
for i, car in enumerate(cars):
    print(i, car)

print("---")

# Нумерация с единицы - для вывода человеку
for number, car in enumerate(cars, start=1):
    print(f"{number}) {car}")

print("---")

# Что именно отдаёт enumerate
print(list(enumerate(cars)))   # [(0, 'bmw'), (1, 'audi'), (2, 'kia')]
````

## 8. `zip` - два списка одновременно

Когда данные лежат в двух разных списках и связаны по позиции
(первый товар - первая цена, второй товар - вторая цена), их перебирают вместе через `zip`:

```python
for товар, цена in zip(products, prices):
```

`zip` берёт по одному элементу из каждого списка и отдаёт их парой.

**Важно:** если списки разной длины, `zip` остановится на самом коротком -
лишние элементы просто не попадут в цикл. Ошибки при этом не будет, поэтому
за длиной списков надо следить самому.

````python
products = ["хлеб", "молоко", "сыр"]
prices = [50, 80, 300]

for product, price in zip(products, prices):
    print(f"{product}: {price} руб.")

print("---")

# Считаем сумму чека
total = 0
for price, count in zip([50, 80, 300], [2, 1, 3]):
    total += price * count
print("Итого:", total)

print("---")

# Списков может быть больше двух
counts = [2, 1, 3]
for product, price, count in zip(products, prices, counts):
    print(f"{product}: {count} шт. по {price} = {price * count}")

print("---")

# Разная длина: лишнее молча отбрасывается
print(list(zip([1, 2, 3], ["a", "b"])))   # [(1, 'a'), (2, 'b')]
````

## 9. Частые ошибки

| Ошибка | Как выглядит | Как правильно |
|---|---|---|
| Нет двоеточия | `for x in nums` -> `SyntaxError` | `for x in nums:` |
| Нет отступа в теле | `IndentationError: expected an indented block` | 4 пробела перед телом |
| Накопитель внутри цикла | `total` обнуляется каждый шаг, ответ неверный | Заводить `total = 0` **до** цикла |
| Забыли `+= 1` в `while` | Ячейка зависает навсегда | Менять переменную условия внутри тела |
| Ждём, что `range(1, 5)` дойдёт до 5 | Последнее число - 4 | `range(1, 6)` |
| `print(range(5))` вместо чисел | Печатает `range(0, 5)` | `list(range(5))` |
| Изменение списка прямо в цикле по нему | Элементы пропускаются | Собирать результат в **новый** список |
| `zip` со списками разной длины | Часть данных молча теряется | Проверять `len()` заранее |
